```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Fetch the CSV path passed by the Julia script.
csv_path = os.environ.get("SIM_CSV_PATH", "CSV data March/default_test_file.csv")
print(f"Loading data from: {csv_path}")

df = pd.read_csv(csv_path)
display(df.head())
```

```python
plt.figure(figsize=(10, 6))
sns.lineplot(data=df, x="Generation", y="population", hue="Rep", palette="tab10")
plt.title("Population over Generations per Replicate")
plt.ylabel("Active Population")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.lineplot(data=df, x="Generation", y="MeanMates", hue="Rep", palette="tab10")
plt.title("Mean Mates per Female over Time")
plt.show()
```
